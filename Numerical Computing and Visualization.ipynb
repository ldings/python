### 1.2 数组的索引与切片

```python
# 10. 一维数组的索引与切片
arr = np.arange(10)
print(arr)
arr_slice = arr[1:7:2]  # 与list不同，切片赋值给变量后，仍关联到数组
print(arr_slice)
arr_slice[0] = 999  # 允许将单个值赋给切片
print(arr)
arr_slice[:] = 888
print(arr)
```

```python
# 11. 拷贝与视图
a = np.arange(12)
b = a  # 完全不拷贝
print(b is a)  # True

c = a.view()  # 视图（浅拷贝）
print(c is a)  # False
print(c.base is a)  # True
c[4] = 999
print(a)

s = a[1:6]  # 切片是一种视图
s[:] = 888
print(a)

d = a.copy()  # 深拷贝
print(d is a)  # False
print(d.base is a)  # False
d[4] = 777
print(a)
```

```python
# 31. 随机漫步1：标准Python方法
import random
position = 0
walk = [position]
nsteps = 1000
for _ in range(nsteps):
    step = 1 if random.randint(0, 1) else -1
    position += step
    walk.append(position)
print(walk)
```

```python
# 32. 随机漫步2：numpy
nsteps = 1000
rng = np.random.default_rng(seed=12345)
draws = rng.integers(0, 2, size=nsteps)
steps = np.where(draws == 0, 1, -1)
walk = steps.cumsum()
walk
```

```python
# 33. 随机漫步3: 统计
print(walk[-1])  # 在1000步后距离原点的步数是多少?
print(walk.min(), walk.max())  # 前后两个方向上离开原点的最远步长是多少？
(np.abs(walk) >= 30).argmax()  # 第一次离开原点30步时，已经走了多少步？
```

```python
# 34. 随机漫步4: 多次模拟
nwalks = 5000
nsteps = 1000
draws = rng.integers(0, 2, size=(nwalks, nsteps))
steps = np.where(draws > 0, 1, -1)
walks = steps.cumsum(axis=1)
walks
```

```python
# 35. 随机漫步5:多次模拟的统计
print((np.abs(walks) >= 30).any(axis=1).sum())  # 有多少次漫步离开原点至少30步？
print((np.abs(walks) >= 30).argmax(axis=1).mean())  # 平均需要走多少步才能离开原点至少30步？
```

```python
# 64. 添加注释和文本
fig, ax = plt.subplots()

t = np.arange(0.0, 5.0, 0.01)
s = np.cos(2*np.pi*t)
line, = ax.plot(t, s, lw=2)

ax.annotate('local max', xy=(2, 1), xytext=(3, 1.5),
            arrowprops=dict(facecolor='black', shrink=0.05))
ax.text(0, -1.5, 'sin function', fontsize=16, weight='bold')
ax.set_ylim(-2, 2)
plt.show()
```
