Q2: 石头剪刀布

```python
import random
choices = ['rock', 'paper', 'scissors']
wins = {'rock': 'scissors', 'scissors': 'paper', 'paper': 'rock'}

while True:
    computer = random.choice(choices)
    user = input("\nEnter rock/paper/scissors (q to quit): ").lower()
    if user == 'q':
        break
        
    if user not in choices:
        print("Invalid input!")
        continue
    print(f"Computer: {computer}")

    if user == computer:
        print("Draw!")
    elif wins[user] == computer:
        print("You win!")
    else:
        print("You lose!")
```

```python
import random
choices = ['rock', 'paper', 'scissors']

while True:
    computer = random.choice(choices)
    user = input("Enter rock / paper / scissors: ").lower()
    if user == 'q':
        break

    if user not in choices:
        print("Invalid input!")
    else:
        print(f"Computer: {computer}")

    if user == computer:
        print("Draw!")
    elif (
        (user == 'rock' and computer == 'scissors') or
        (user == 'paper' and computer == 'rock') or
        (user == 'scissors' and computer == 'paper')
    ):
        print("You win!")
    else:
        print("You lose!")
```

Q3: Fibonacci数列

```python
def fib(n):  # 普通函数
    i, a, b = 0, 0, 1
    while i < n:
        a, b = b, a + b  # 交换变量
        i += 1
    return b

print([fib(i) for i in range(10)])  # 普通函数每次调用都会从头开始执行
```

```python
def fib2(n):  # 生成器
    i, a, b = 0, 0, 1
    while i < n:
        yield b  # 生成器函数在yield处返回，并挂起当前状态，下次执行时由此进入
        a, b = b, a + b
        i += 1
        
print(list(fib2(10)))
```

Q4: 银行账号

```python
class Account:
    account_type = 'bank account'

    def __init__(self, account_number, account_holder, opening_balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.opening_balance = opening_balance

    def deposit(self, amount):
        if amount > 0:
            self.opening_balance += amount

    def withdraw(self, amount):
        if 0 < amount < self.opening_balance:
            self.opening_balance -= amount

    def get_balance(self):
        return self.opening_balance

    def __str__(self):
        return f'Account[{self.account_number}]{self.account_holder}, current account={self.opening_balance}'
    
    def __add__(self, another_account):
        if self.account_holder == another_account.account_holder:
            self.opening_balance += another_account.opening_balance
        return self

class CurrentAccount(Account):
    def __init__(self, account_number, account_holder, opening_balance, overdraft_limit):
        super().__init__(account_number, account_holder, opening_balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        if self.opening_balance - amount >= self.overdraft_limit:
            self.opening_balance -= amount

acc1 = Account('1001', 'Bella', 12.45)
acc2 = Account('1099', 'Bella', 10.50)
print(acc1 + acc2)
```

Q5: 医院急诊记录

```python
file_path = 'ER_XIAN.log'
file = open(file_path, 'r')
lines = file.readlines()
for line in lines:
    fields = line.split(':')
    name = fields[1]
    gender = fields[2]
    if not name:
        if gender == 'Male':
            print('JOHN DOE')
        elif gender == 'female':
            print('JANE DOE')
    else:
        print(name.upper())
file.close()
```
