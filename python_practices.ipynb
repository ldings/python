 1. 使用函数编写一段程序，要求用户输入一句英文，将其中的单词反向排列并输出。例如：用户输入 My name is Michele，则输出 Michele is name My。

```python
def reverseWord(w):
    return ' '.join(w.split()[::-1])

reverseWord('My name is Michele')
```

2. 编写一个程序，要求用户输入一个数字，输出这个数字所有的除数（某个数的除数是指能够整除它的数字，如13是26个一个除数，因为26/13的余数为0）

```python
num = int(input("Please choose a number to divide: "))

divisorList = []
for i in range(1, num + 1):
    if num % i == 0:
        divisorList.append(i)

print(divisorList)
```

3. 编写一个石头剪刀布游戏，系统随机生成石头、剪刀、布的一种，要求玩家输入（使用input函数）其中一种，比较并输出游戏结果，并询问是否继续新的游戏

```python
import random

options = ["石头", "剪刀", "布"]
while True:
    computer_choice = random.choice(options)    
    player_choice = input("请输入 石头、剪刀 或 布（输入 q 退出）：").strip()
    
    if player_choice == "q":
        print("游戏结束，感谢游玩！")
        break

    if player_choice not in options:
        print("❌ 输入无效，请重新输入！")
        continue

    print(f"电脑出了：{computer_choice}")
    print(f"你出了：{player_choice}")

    if player_choice == computer_choice:
        print("🤝 平局！")
    elif (player_choice == "石头" and computer_choice == "剪刀") \
        or (player_choice == "剪刀" and computer_choice == "布") \
        or (player_choice == "布" and computer_choice == "石头"):
        print("🎉 你赢了！")
    else:
        print("😢 你输了！")

    again = input("是否继续游戏？(y/n)：").strip().lower()
    if again != "y":
        print("游戏结束，再见！")
        break
```

4. 随机生成1到9之间的数字（包含1和9），要求用户猜这个数字。当用户输入错误时，提示比该数字大还是小；当用户输入 exit 时退出游戏；当用户猜中时提示成功，显示用户猜了几次，并询问是否继续游戏。

```python
import random
rd = random.randint(1,30)
guess = 0
c = 0
while guess != rd and guess != "exit":
    guess = input("Enter a guess between 1 to 9")  

    try:
        guess = int(guess)
        c += 1

        if guess < rd:
            print(guess, ": Too low")
        elif guess > rd:
            print(guess, ": Too high")
        else:
            print(guess, ": Right!")
            print("You took only", c, "tries!")
    except:
        break
```

5. 编写一个 Cows and bulls 游戏，游戏的玩法如下：
随机生成一个4位数字（每位数字各不相同），要求玩家猜这个数字；每次玩家猜测时，若对应的位置为正确的数字，称为cow，若正确的数字在错误的位置，称为bull；提示玩家cow和bull的数量。
例如，随机产生的数字为1038，若玩家猜测 1234，提示 2 cows, 0 bulls，若玩家猜测 1356，提示 1 cow, 1 bull

```python
import random
def compare_numbers(number, user_guess):
    cows, bulls = 0, 0
    for i in range(len(user_guess)):
        if number[i] == user_guess[i]:
            bulls += 1
        if user_guess[i] in number:
            cows += 1
    return cows, bulls

number = random.sample(range(10), 4)
number = ''.join([str(i) for i in number])
guesses = 0

while True:
    user_guess = input("Give me your best guess!")

    if user_guess == '?':
        print("The number is", number)
        break

    if user_guess == 'exit':
        break
    
    try:
        user_guess = f'{int(user_guess):04d}'
        assert(len(set(user_guess)) == 4)
        c, b = compare_numbers(number, user_guess)
        guesses += 1
        if b == 4:
            print(f"You win the game in {guesses} times! The number is {number}")
            break
        else:
            print(user_guess, f": You have {c} cows, and {b} bulls.")
    except:
        print('Please give four different digits')
        continue
```
