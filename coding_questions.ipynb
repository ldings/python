17. 阶乘指数列1, 2, 6, 24, 120, 720, ...，写出阶乘的生成器函数。

````python
def f(n):
    i, j = 1, 1
    while i <= n:
        j *= i
        yield j
        i += 1
````

18. 斐波那契数列是一个形如0, 1, 1, 2, 3, 5, 8, 13...的数列，补全斐波那契数列的生成器。

````python
def fib(n):
    i, a, b = 0, 0, 1
    while i < n:
        yield b
        a, b = b, a + b
        i += 1
````

23. 编写一个表示圆形的类Circular，包含一个实例变量radius，表示圆形的半径，在实例化时将初始化radius变量。该类是可求长度的，其长度即为圆形的周长。

````python
import math
class Circular:
    def __init__(self, r):
        self.radius = r
    
    def __len__(self):
        return 2 * math.pi * self.radius
````

24. 编写一个表示矩形的类Rectangle，包含两个实例变量length和width，分别表示矩形的长和宽，该类是可求长度的，其长度即为矩形的周长。

````python
class Rectangle:
    def __init__(self, l, w):
        self.length = l
        self.width = w
        
    def __len__(self):
        return 2 * self.length + 2 * self.width
````

26. 比特币使用的椭圆曲线称为secp256k1，其方程为：$y^2=x^3+7$。编写一个类Point表示该椭圆曲线上的点：（1）该类包含两个实例变量x和y。（2）在实例化时，该类接受两个参数分别初始化这两个变量的值，同时检查初始化的值对应的点是否在曲线上，若不在曲线上，则抛出ValueError异常。（3）该类还允许判断两个点是否相等，当两个点的x和y值分别相等时才返回True。

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        if self.y ** 2 != self.x ** 3 + 7:
            raise ValueError()
        
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
````

（注：圆周率为math.pi）

````python
import math
class NonIntersectingLimacon:
    def __init__(self, a, b):
        if abs(a) <= abs(b):
            raise ValueError
        self.a = a
        self.b = b
    
    def __len__(self):
        return round(2 * math.pi * self.a + math.pi * self.b ** 2 / self.a)

l = NonIntersectingLimacon(4, 3)
len(l)
````
